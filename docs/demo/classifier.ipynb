# Training clip classifier demo

Plays each training audio clip in `audio-uploader/training-uploads/` and shows the environment label predicted by `environment-classifier/audio_analysis/classifier.py`.

```python
import sys
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "environment-classifier").is_dir():
            return candidate
    raise RuntimeError("Could not locate repo root (looked for an 'environment-classifier' directory)")


REPO_ROOT = find_repo_root(Path.cwd())
sys.path.insert(0, str(REPO_ROOT / "environment-classifier"))

from audio_analysis import analyze_file
from IPython.display import Audio, display
```

```python
TRAINING_DIR = REPO_ROOT / "audio-uploader" / "training-uploads"
training_files = sorted(TRAINING_DIR.glob("*.wav"))

if not training_files:
    print(f"No training .wav files found in {TRAINING_DIR}")
else:
    print(f"Found {len(training_files)} training file(s) in {TRAINING_DIR}")
```

```python
for path in training_files:
    print(f"\n=== {path.name} ===")
    display(Audio(filename=str(path)))  # plays the original file as-is

    features, result = analyze_file(path)  # librosa.load -> extract_features -> classify

    print(f"Environment: {result.environment.value}")
    print(f"Reasoning:   {result.reasoning}")
```
